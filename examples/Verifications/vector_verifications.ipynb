```python
import numpy as np
from hypercomplex_engine.core.vectorized import VectorizedStandard
from hypercomplex_engine import build_table

vec = VectorizedStandard()
rng = np.random.default_rng(0)
for n in range(2, 11):
    N = 1 << n
    signs, idx = build_table("standard", n)
    for trial in range(200):
        for dens in (0.05, 0.3, 1.0):           # sparse -> dense
            x = rng.normal(size=N) * (rng.random(N) < dens)
            y = rng.normal(size=N) * (rng.random(N) < dens)
            got = vec.multiply_arrays(x, y)
            ref = np.zeros(N)
            S = signs[:len(x), :len(y)] * np.outer(x, y)
            np.add.at(ref, idx[:len(x), :len(y)].ravel(), S.ravel())
            assert np.allclose(got, ref, atol=1e-12), (n, dens)
print("vectorized sparse multiplier: all cross-checks passed")
```
