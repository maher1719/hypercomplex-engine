```python
import time
import random
import sys

# Ensure Python can handle high-digit serialization strings for printing text
sys.set_int_max_str_digits(2_000_000)

try:
    from hypercomplex import multiply, build_table
except ImportError:
    print("❌ Error: 'hypercomplex-engine' package not found.")
    print("Please make sure you are running this script inside your repository directory")
    print("or run 'pip install hypercomplex-engine' first.")
    sys.exit(1)

def run_comprehensive_benchmark():
    print("=" * 80)
    print(" 🌋      HYPERCOMPLEX-ENGINE ARCHITECTURAL AMMUNITION SUITE METRICS      🌋 ")
    print("=" * 80)
    
    # ------------------------------------------------------------------------
    # STAGE 1: The Setup Bottleneck Test (Dimension Exponent n=12 -> 4,096 Dim)
    # ------------------------------------------------------------------------
    n_medium = 12
    print(f"\n[STAGE 1] Testing Single-Operation Setup + Execution (n={n_medium} | 4,096 Dimensions)")
    print("-" * 80)
    
    # Approach A: Traditional Memory Matrix
    print("  ▪ Traditional: Allocating memory arrays and building structural tables...")
    start_t = time.perf_counter()
    signs, indices = build_table("standard", n=n_medium)
    # Simulate a single matrix lookup call
    res_sign = signs[5][10]
    res_idx = indices[5][10]
    t_traditional = time.perf_counter() - start_t
    print(f"    Total Traditional Time (Setup + Lookup): {t_traditional:.6f} seconds")
    
    # Approach B: O(1) Fast Engine
    print("  ▪ Maher's O(1) Fast Engine: Computing sign on the fly...")
    start_t = time.perf_counter()
    res_fast_single = multiply("standard", (1, 5), (1, 10), engine="fast")
    t_fast_single = time.perf_counter() - start_t
    print(f"    Total Fast Engine Time:                  {t_fast_single:.6f} seconds")
    
    # Approach C: Holographic Engine
    print("  ▪ Maher's Holographic Engine: Descending fractal blocks...")
    start_t = time.perf_counter()
    res_holo_single = multiply("standard", (1, 5), (1, 10), engine="holographic")
    t_holo_single = time.perf_counter() - start_t
    print(f"    Total Holographic Time:                  {t_holo_single:.6f} seconds")
    
    print(f"\n  🚀 For ad-hoc operations, Fast Engine is {t_traditional / t_fast_single:.2f}x faster.")
    print(f"  🚀 For ad-hoc operations, Holographic is {t_traditional / t_holo_single:.2f}x faster.")

    # ------------------------------------------------------------------------
    # STAGE 2: Head-to-Head Multi-Iteration Engine Loops (50,000 Iterations)
    # ------------------------------------------------------------------------
    iterations = 50_000
    print(f"\n[STAGE 2] Head-to-Head Engine Loops ({iterations:,} iterations per test)")
    print("-" * 80)
    
    # Test 2A: Structural Indices (Powers of 2 | Favors Holographic boundary short-circuits)
    print("  Running Test 2A: Structural Boundary Pillars (e_1024 × e_2048)...")
    structural_pairs = [((1, 1024), (1, 2048)) for _ in range(iterations)]
    
    start_t = time.perf_counter()
    for a, b in structural_pairs:
        multiply("standard", a, b, engine="fast")
    loop_fast_struct = time.perf_counter() - start_t
    
    start_t = time.perf_counter()
    for a, b in structural_pairs:
        multiply("standard", a, b, engine="holographic")
    loop_holo_struct = time.perf_counter() - start_t
    
    print(f"    ▪ O(1) Fast Engine Loops:       {loop_fast_struct:.5f} seconds")
    print(f"    ▪ Holographic Engine Loops:     {loop_holo_struct:.5f} seconds")
    if loop_holo_struct < loop_fast_struct:
        print(f"    👉 Holo wins by {loop_fast_struct / loop_holo_struct:.2f}x via 'is_structural' boundary triggers!")
    else:
        print(f"    👉 O(1) Fast wins by {loop_holo_struct / loop_fast_struct:.2f}x!")

    # Test 2B: Dense Complex Indices (Chaotic odd numbers deep inside space)
    print("\n  Running Test 2B: Dense Random Odd Indices (Fractal Matrix Chaos)...")
    dense_pairs = [((1, random.randint(3001, 4095) | 1), (1, random.randint(3001, 4095) | 1)) for _ in range(iterations)]
    
    start_t = time.perf_counter()
    for a, b in dense_pairs:
        multiply("standard", a, b, engine="fast")
    loop_fast_dense = time.perf_counter() - start_t
    
    start_t = time.perf_counter()
    for a, b in dense_pairs:
        multiply("standard", a, b, engine="holographic")
    loop_holo_dense = time.perf_counter() - start_t
    
    print(f"    ▪ O(1) Fast Engine Loops:       {loop_fast_dense:.5f} seconds")
    print(f"    ▪ Holographic Engine Loops:     {loop_holo_dense:.5f} seconds")
    if loop_fast_dense < loop_holo_dense:
        print(f"    👉 O(1) Fast wins by {loop_holo_dense / loop_fast_dense:.2f}x! Bitwise logic outruns deep recursion stack branches.")
    else:
        print(f"    👉 Holo wins by {loop_fast_dense / loop_holo_dense:.2f}x!")

    # ------------------------------------------------------------------------
    # STAGE 3: Extreme Scale Space Expansion (The Monster Bignum Profiles)
    # ------------------------------------------------------------------------
    print("\n[STAGE 3] Extreme Scale Space Expansion (Monster Big-Integer Tests)")
    print("-" * 80)
    
    digit_targets = [4300, 60000, 1000000]
    
    for digits in digit_targets:
        print(f"\n  🔥 Triggering Challenge Space: {digits:,}-Digit Index Vectors")
        
        # Native programmatic generation of large odd digits to prevent layout short-circuits
        huge_int_a = int("9" * (digits - 1) + "7")
        huge_int_b = int("8" * (digits - 1) + "3")
        
        huge_a = (1, huge_int_a)
        huge_b = (1, huge_int_b)
        
        # Test 1: Fast bitwise engine
        start_t = time.perf_counter()
        res_fast = multiply("standard", huge_a, huge_b, engine="fast")
        t_fast = time.perf_counter() - start_t
        print(f"    ▪ O(1) Fast Engine Runtime:   {t_fast:.6f} seconds")
        
        # Test 2: Holographic recursive engine
        start_t = time.perf_counter()
        res_holo = multiply("standard", huge_a, huge_b, engine="holographic")
        t_holo = time.perf_counter() - start_t
        print(f"    ▪ Holographic Engine Runtime: {t_holo:.6f} seconds")
        
        # Integrity verification step
        validation_match = (res_fast == res_holo)
        print(f"    👉 Cross-Engine Exact Equality Verified? {validation_match}")
        
        if t_fast > 0:
            print(f"    👉 Speed Divergence Factor:              {t_holo / t_fast:.2f}x faster via bitwise laws")

    print("\n" + "=" * 80)
    print("                      🏁 BENCHMARK EVALUATION COMPLETE 🏁                     ")
    print("=" * 80)

if __name__ == "__main__":
    run_comprehensive_benchmark()
```
